# Extracción de datos

### Importación de librerías

```python
import openeo
import numpy as np
import rasterio
from datetime import datetime, timedelta
from pathlib import Path
```

### Información para la extracción de datos

```python
conexion = openeo.connect(url="https://openeo.dataspace.copernicus.eu").authenticate_oidc()

lagos = {
    "atitlan": {
        
        "bbox": {"west": -91.326256, "east": -91.071510, "south": 14.594800, "north": 14.750979,
                 "crs": "EPSG:4326"},
        "fechas": ["2025-01-18", "2025-04-13", "2025-05-13", "2025-07-17", "2025-11-21", "2025-12-29",
                   "2026-02-12", "2026-03-24", "2026-04-13", "2026-04-28", "2026-07-22"]},
    "amatitlan": {"bbox": {"west": -90.638065, "east": -90.512924, "south": 14.412347,
                           "north": 14.493799,
                           "crs": "EPSG:4326"},
        "fechas": ["2025-01-28", "2025-04-15", "2025-04-28", "2025-11-24", "2026-01-08", "2026-02-02",
                   "2026-02-07", "2026-03-29", "2026-04-13", "2026-04-28", "2026-06-19"]},
        }

BANDAS = ["B03", "B04", "B05", "B08"]
DIRECTORIO_SALIDA = Path("datos_lagos")
```

```python
bandas = ["B03", "B04", "B05", "B08"]
for nombre_lago, configuracion in lagos.items():
    carpeta_lago = Path("datos_lagos") / nombre_lago
    carpeta_lago.mkdir(parents=True, exist_ok=True)
    for fecha in configuracion["fechas"]:
        fecha_siguiente = (datetime.fromisoformat(fecha) + timedelta(days=1)).date().isoformat()
        for banda in bandas:
            try:
                cubo = conexion.load_collection(collection_id="SENTINEL2_L1C", spatial_extent=configuracion["bbox"], temporal_extent=[fecha, fecha_siguiente], bands=[banda])
                imagen = cubo.resample_spatial(resolution=20, projection=32615, method="bilinear").reduce_dimension(dimension="t", reducer="first")
                imagen.download(outputfile=str(carpeta_lago / f"{nombre_lago}_{fecha}_{banda}.tif"), format="GTiff")
                print(f"{nombre_lago} - {fecha} - {banda}: descargada")
            except Exception as error:
                print(f"{nombre_lago} - {fecha} - {banda}: {error}")
```
